```python
#BallScrewKneeVectors_v2B
#Code to compute knee torque when using a ball screw actuator with lever arm. 
#Vector analysis
#v2 mistake in code for computing the knee torque.  component of ballscrew force along lever arm not being computed.
#v2B Component of ballscrew force along lever arm computed. 

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from pathlib import Path
from vectoroperations import rotate2dsweep
```

```python
plt.close('all')   
```

```python

# %%
O=np.array([0,0]) #Knee center
P0=np.array([0,0.1]) #Patella or lever arm middle position
#AngP = np.linspace(135,210, 12)  # 181 points, evenly spaced -90 to 90 (step = 1)
AngP=-45
P = rotate2dsweep(P0, AngP)   # anglein degrees
H_bscrew=0.10 # Height of ball screw above knee center.
Forceballscrew=5000 #Force applied by ball screw in N
L_pushrod=0.21 #Pushrod length from ballscrew


# Rotate2D(vector, angle in degrees CW +)
F = rotate2dsweep(P, 170)*2   # foot vector

Pvert = np.array([P[0], np.full_like(P[0], H_bscrew)]) # Vectors from knee to vertical projection of P onto line of action of ball screw.
Pup=Pvert-P #Vector from P to Pvert 
```

```python

# %% calculating point B (ball screw) using trianlge of B, P, and Pup

L_lever = np.linalg.norm(P0)   # Length of lever arm. 
L_Pup=np.linalg.norm(Pup, axis=0) #Distance from P to Pvert
L_pushrodmin = np.sqrt(L_lever**2 + H_bscrew**2)  #checking pushrod length possiilities

L_bscrew=np.sqrt(L_pushrod**2-L_Pup**2) #Distance from Pup to B (ball screw)

BSmag = np.array([-L_bscrew, np.zeros_like(L_bscrew)]) #Ball screw vector
BS=BSmag+Pvert #Ball screw position in global coordinates
BSvec=Pvert-BS #Vector representing ball screw axis


# print('P0:', P0)
# print('P:', P)  
# print('Pup:', Pup)
# print('F:', F)
# print('BS:', BS)
# print('L_Pup:', L_Pup)
# print('L_bscrew:', L_bscrew)
# print('L_pushrodmin:', L_pushrodmin)

```

```python
# %% Torque calculations




# calculations for a single lever arm.
Pushrodvec=P - BS
Pushrodvec_unit = Pushrodvec / np.linalg.norm(Pushrodvec, axis=0)
BSvec_unit = BSvec / np.linalg.norm(BSvec, axis=0)
cosval=np.sum(BSvec_unit * Pushrodvec_unit, axis=0)
#Fpushrod = Forceballscrew*cosval  # finding componnt of ballscrew force acting along pushrod
Fpushrod = Forceballscrew/cosval  # finding componnt of ballscrew force acting along pushrod
torque_O = Fpushrod * np.cross(P, Pushrodvec_unit, axis=0) # Torque at knee center O



# print('Torque at knee center O:', torque_O, "N.m")


# df = pd.DataFrame({
#     'Angle_deg': AngP,
#     'Torque_Nm': torque_O
# })
# df.to_csv('Ballscrewknee_torque_vs_angle.csv', index=False)


```
